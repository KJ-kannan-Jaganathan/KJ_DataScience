```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm

# Make plots a little bigger
plt.rcParams['figure.figsize'] = (7, 4)
plt.rcParams['axes.grid'] = True

# Reproducible randomness
np.random.seed(11)

# Number of people in each group
n_per_group = 35

# Group labels and their true means (try changing these means)
groups = ['A','B','C','D']
true_means = [70, 72, 78, 82]   # change to [75,75,75,75] and see what happens!
within_sd  = 8                  # within-group standard deviation

# Build a DataFrame with generated data
data_frames = []
for g, mu in zip(groups, true_means):
    vals = np.random.normal(loc=mu, scale=within_sd, size=n_per_group)
    data_frames.append(pd.DataFrame({'group': g, 'score': vals}))

df = pd.concat(data_frames, ignore_index=True)
df.head()
```

Ho = Are all group means equal?

Ha = Atleast one group mean is different

```python
alpha = 0.05

samples = [df.loc[df.group == g,'score'].values for g in groups]
F, p = stats.f_oneway(*samples) #* argument unpacking proerations
```

```python
samples
```

```python
if p < alpha:
    print('Reject Ho')
else:
    print('Fail to reject Ho')
```

```python
astronaut = imgs['astronaut'] # 3D RGB
bright = np.clip(astronaut + 0.4, 0, 1) #Np.clip lower, upper
show(astronaut,"OriginalImage")
show(bright,"Brightened Image")
```

```python
astronaut = imgs['astronaut'] # 3D RGB
dark = np.clip(astronaut - 0.20, 0, 1) #Np.clip - restricts the values between lower limit, upper limit
show(astronaut,"OriginalImage")
show(dark,"Darkened Image")
```

Brightness is the overall lightness or darkness of an entire image, affecting all its tones uniformly. Contrast, on the other hand, is the difference between the darkest and lightest parts of an image, altering the visual separation between light and dark areas without changing their individual brightness.

```python
contrast = np.clip(1.3*astronaut,0,1)
show(astronaut,"OriginalImage")
show(contrast,"Increased Contrast")
```
